```python
print("First 5 rows:")
display(df.head())

print("Dataset size (rows, columns):")
print(df.shape)

print("Column names:")
print(df.columns)

print("Data types and non-null values:")
df.info()

print("Statistical summary:")
display(df.describe(include="all"))

print("Missing values per column:")
print(df.isnull().sum())

print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Make a separate copy for cleaning
clean_df = df.copy()

# Remove duplicate complete rows
clean_df = clean_df.drop_duplicates()

# Remove extra spaces from text columns
text_columns = ["Customer", "Product", "City"]

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()

# Make city names consistent
clean_df["City"] = clean_df["City"].str.title()

# Convert numerical columns to numeric values
clean_df["Quantity"] = pd.to_numeric(
    clean_df["Quantity"], errors="coerce"
)

clean_df["Price"] = pd.to_numeric(
    clean_df["Price"], errors="coerce"
)

# Convert order dates into date format
clean_df["Date"] = pd.to_datetime(
    clean_df["Date"], errors="coerce"
)

# Fill missing customer names and cities with "Unknown"
clean_df["Customer"] = clean_df["Customer"].fillna("Unknown")
clean_df["City"] = clean_df["City"].fillna("Unknown")

# Remove rows with missing essential numerical values
clean_df = clean_df.dropna(
    subset=["Quantity", "Price", "Date"]
)

# Calculate total sales for each order
clean_df["Sales"] = clean_df["Quantity"] * clean_df["Price"]

# Save the cleaned data
clean_df.to_csv("sales_cleaned.csv", index=False)

print("Cleaning completed!")
print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))

display(clean_df.head())
```

```python
print("Remaining missing values:")
print(clean_df.isnull().sum())

print("Remaining duplicate rows:")
print(clean_df.duplicated().sum())

print("Final data types:")
print(clean_df.dtypes)

print("Total sales across all orders:")
print(clean_df["Sales"].sum())
```
